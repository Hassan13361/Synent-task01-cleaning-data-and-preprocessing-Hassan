```python
# Task No:1
import seaborn as sns
import pandas as pd
titanic_data=sns.load_dataset("titanic")
```

```python
#Titanic Data set stored in data frame
df_td= pd.DataFrame(titanic_data)
df_td.head()
```

```python
# Checking Missing Values in terms of percentage
missing_per=(df_td.isnull().sum().sort_values(ascending=False)*100)/len(df_td)
print(missing_per)
```

```python
#Handling Missing Values
# with less missing values of embarked and embark so deleting null of those rows is safr
embarked_del_rows=df_td.dropna(subset=['embarked','embark_town'])
print("Deleted rows:",len(embarked_del_rows))
# with only 19% of missing data in age we take the mean and store it at the null values
avg_val=round(df_td['age'].mean())
df_td['age']=df_td['age'].fillna(avg_val)
# with over 30% missing values it is safe to delete the those colums
df_td.drop(columns='deck')
```

```python
# Now we convert the datatype of the columns age to int and rename the columns sex to gender
df_td['age']=df_td['age'].astype(int)
df_td.rename(columns={'sex':'Gender'},inplace=True)
df_td
```
